# Theory Questions

- 1.Difference between a Function and a Method in Python:
  
  Function is defined using def keyword and is independent.

  Method is associated with an object (e.g., list, string).

  Example:

 def add(x, y):
    return x + y

 print("hello".upper())  # `upper()` is a method of string


- 2.Function Arguments and Parameters:

  Parameters are variables in function definition.

  Arguments are actual values passed.

  Example:


    def greet(name):  # name is a parameter
    print(f"Hello, {name}")

    greet("Alice")  # "Alice" is an argument

- 3. Ways to Define and Call a Function:
Normal function:

   def square(x):
    
    return x * x
    
    square(5)
    
    Lambda function:

    square = lambda x: x * x
    
    square(5)

- 4.Purpose of return Statement:
To send result back to the caller and exit function.

 Example:


  def multiply(a, b):
    
    return a * b

- 5.Iterators vs Iterables:

 Iterable: Object you can iterate over (e.g., list, string).

 Iterator: Object with __next__() and __iter__().

  Example:


  nums = [1, 2, 3]  # Iterable
  
  it = iter(nums)  # Iterator
  
  next(it)

- 6. Generators in Python:
Functions that yield values one at a time using yield.

 Example:


 def countdown(n):
    
    while n > 0:
        
        yield n
        
        n -= 1

- 7.Advantages of Generators:

 Memory efficient (lazy evaluation)


 Faster for large data


 Simplifies code using yield

- 8.Lambda Function and Typical Use:

 Anonymous one-line function. Example:


  
 square = lambda x: x * x

  Used in map, filter, or quick operations.

- 9.Purpose and Usage of map():

 Applies a function to all items in an iterable.

 Example:


  nums = [1, 2, 3]
  
  squares = list(map(lambda x: x*x, nums))


- 10.Difference between map(), reduce(), filter():

  map(): Transforms each item.


  filter(): Filters items by condition.

  reduce(): Aggregates values to a single result.

  Example:


  from functools import reduce

  nums = [1, 2, 3, 4]

  reduce(lambda x, y: x + y, nums)  # Output: 10

# Practical Questions

- 1.Sum of all even numbers in a list

```python
def sum_of_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)

# Example usage
my_list = [2, 4, 6, 8, 10, 12]
print(sum_of_even_numbers(my_list))
```

- 2.Reverse a string

```python
def reverse_string(s):
    # This function returns the reverse of the input string
    return s[::-1]

# Example usage
text = "hello"
reversed_text = reverse_string(text)
print("Original string:", text)
print("Reversed string:", reversed_text)
```

- 3.Square each element in a list

```python
def square_numbers(numbers):
    # This function returns a new list with the squares of each number in the input list
    return [num ** 2 for num in numbers]

# Example usage
my_list = [1, 2, 3, 4, 5]
squared_list = square_numbers(my_list)
print("Original list:", my_list)
print("Squared list:", squared_list)
```

- 4.Check if a number is prime (1–200)

```python
def is_prime(n):
    # Check for valid range
    if n < 2 or n > 200:
        return False

    # Check for factors
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage
for number in range(1, 201):
    if is_prime(number):
        print(f"{number} is a prime number.")
```

- 5.Fibonacci sequence generator up to n terms

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            fib = self.a + self.b
            self.a, self.b = self.b, fib
            self.count += 1
            return fib

# Example usage
fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

- 6.Generator for powers of 2 up to n

```python
def powers_of_two(max_exponent):
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent

# Example usage
for value in powers_of_two(5):
    print(value)
```

- 8.Sort list of tuples by second element

```python
# List of tuples
tuples_list = [(1, 4), (3, 1), (5, 9), (2, 7)]

# Sorting the list of tuples based on the second element of each tuple
sorted_list = sorted(tuples_list, key=lambda x: x[1])

# Printing the sorted list
print(sorted_list)
```

- 9.Convert Celsius to Fahrenheit using map()

```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 37, 100, -5]

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Using map() to apply the conversion function to each element in the list
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Printing the converted temperatures
print(fahrenheit_temps)
```

- 10.Remove vowels using filter()

```python
# Function to check if a character is not a vowel
def is_not_vowel(char):
    vowels = 'aeiouAEIOU'
    return char not in vowels

# Given string
input_string = "Hello, World!"

# Using filter() to remove vowels from the string
filtered_string = ''.join(filter(is_not_vowel, input_string))

# Printing the string after removing vowels
print(filtered_string)
```

- 11.Accounting utility with lambda and map

```python
orders = [
    [34587, "Learning Python, Mark Lutz", 4, 40.95],
    [98762, "Programming Python, Mark Lutz", 5, 56.80],
    [77226, "Head First Python, Paul Barry", 3, 32.95],
    [88112, "Einführung in Python3, Bernd Klein", 3, 24.99]
]

# Using lambda and map
invoice_totals = list(map(lambda order: (
    order[0],
    round(order[2] * order[3] if order[2] * order[3] >= 100 else order[2] * order[3] + 10, 2)
), orders))

print(invoice_totals)
```
